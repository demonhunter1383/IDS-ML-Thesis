```python
import pandas as pd
import numpy as np
import os

constant_cols_all = {}

for filename in sorted(os.listdir(data_dir)):
    if filename.endswith(".csv"):
        file_path = os.path.join(data_dir, filename)
        try:
            df_temp = pd.read_csv(file_path)
        except UnicodeDecodeError:
            df_temp = pd.read_csv(file_path, encoding="latin1")

        df_temp.columns = df_temp.columns.str.strip()

        # Is "Fwd Header Length.1" actually a byte-for-byte duplicate
        # of "Fwd Header Length", or just a similar name?
        if "Fwd Header Length.1" in df_temp.columns and "Fwd Header Length" in df_temp.columns:
            identical = df_temp["Fwd Header Length.1"].equals(df_temp["Fwd Header Length"])
            print(f"{filename}: duplicate of Fwd Header Length? {identical}")

        # Columns where every value is the same carry zero
        # information for a classifier -- find them
        numeric = df_temp.select_dtypes(include="number")
        constant_cols = numeric.columns[numeric.nunique(dropna=False) <= 1].tolist()
        if constant_cols:
            constant_cols_all[filename] = constant_cols

print()
print("Zero-variance columns per file:")
for filename, cols in constant_cols_all.items():
    print(filename, "->", cols)
```

## Legacy preprocessing pipeline (DO NOT RUN)

Cells 16 to 18 are kept for historical reference. They write to `data/processed/clean_flows.parquet`, an existing dataset that must not be overwritten accidentally.

The current identified-flow preparation script is `src/prepare_identified_flows.py`. Its output is `data/processed/identified_flows.parquet`.

**Do not execute cells 16 to 18.**

```python
import pandas as pd
import numpy as np
import os

# your existing data_dir line goes here, e.g.:
# data_dir = "data/raw"

DROP_COLUMNS = [
    "Fwd Header Length.1",
    "Bwd PSH Flags", "Fwd URG Flags", "Bwd URG Flags", "CWE Flag Count",
    "Fwd Avg Bytes/Bulk", "Fwd Avg Packets/Bulk", "Fwd Avg Bulk Rate",
    "Bwd Avg Bytes/Bulk", "Bwd Avg Packets/Bulk", "Bwd Avg Bulk Rate",
]

def load_and_clean_file(file_path):
    try:
        df = pd.read_csv(file_path)
    except UnicodeDecodeError:
        df = pd.read_csv(file_path, encoding="latin1")

    df.columns = df.columns.str.strip()
    df["Label"] = df["Label"].str.replace("\ufffd", "-", regex=False)

    df = df.drop(columns=[c for c in DROP_COLUMNS if c in df.columns])

    numeric = df.select_dtypes(include="number").replace([np.inf, -np.inf], np.nan)
    keep_mask = ~numeric.isna().any(axis=1)
    df = df[keep_mask]

    return df

frames = []
for filename in sorted(os.listdir(data_dir)):
    if filename.endswith(".csv"):
        file_path = os.path.join(data_dir, filename)
        frames.append(load_and_clean_file(file_path))

df_all = pd.concat(frames, ignore_index=True)

print("Shape check:", df_all.shape, "-- expect (2827876, 68)")
```

```python
numeric_cols = df_all.select_dtypes(include="number").columns
df_all[numeric_cols] = df_all[numeric_cols].astype("float32")

os.makedirs("data/processed", exist_ok=True)
df_all.to_parquet("data/processed/clean_flows.parquet", index=False)

size_mb = os.path.getsize("data/processed/clean_flows.parquet") / 1_000_000
print(f"Saved. File size: {size_mb:.1f} MB")
```
